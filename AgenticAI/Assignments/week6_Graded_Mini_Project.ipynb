```python
#Section A — LLM Foundations & Hugging Face
#1. Environment Setup & Model Loading
!pip install transformers --quiet
from transformers import pipeline
```

```python
generator = pipeline("text-generation", model="distilgpt2")
```

```python
prompt = "AI is transforming industries by"

outputs = generator(prompt, max_length=40, num_return_sequences=3)

for i, out in enumerate(outputs):
    print(f"\n--- Output {i+1} ---\n")
    print(out['generated_text'])
```

```python
#2. Tokenisation Demo
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("distilgpt2")

sentence = "LLMs are powerful tools for natural language understanding."

tokens = tokenizer.tokenize(sentence)
token_ids = tokenizer.encode(sentence)
sequence_length = len(token_ids)

print("Tokens:", tokens)
print("Token IDs:", token_ids)
print("Sequence Length:", sequence_length)
```

```python
#Section B — Prompt Engineering
#Prompt Tuning Challenge
#Summarisation Prompt
prompt_sum = "Summarise in 30 words: Artificial Intelligence is changing sectors like healthcare, retail, and manufacturing by automating tasks and enabling smarter decision-making."

print(generator(prompt_sum, max_length=60)[0]['generated_text'])
```

```python
#Q&A Prompt
prompt_qa = "Question: What is the capital of Japan? Answer:"

print(generator(prompt_qa, max_length=40)[0]['generated_text'])
```

```python
#Creative Prompt
prompt_poem = "Write a 4-line poem about AI:"

print(generator(prompt_poem, max_length=50)[0]['generated_text'])
```

```python
'''
Reflection

The prompt variations clearly affected how the model responded. When the prompt was more specific such as “Summarise in 30 words”  the output became shorter, structured, and more factual. For Q&A prompts, adding the word “Answer:” guided the model to produce direct responses rather than long explanations. Creative prompts, especially those specifying a poem format, pushed the model toward rhythmic and imaginative text. Small changes in phrasing significantly changed the tone, detail, and structure of the output. The model behaved more predictably when constraints were explicitly stated. This shows how important prompt clarity is when working with LLMs vague prompts lead to vague outputs, while precise instructions generate focused and useful results.
'''
```

```python
#Section C — Embeddings with Gensim
#Word Embeddings (GloVe)
!pip install gensim --quiet
import gensim.downloader as api

glove = api.load("glove-wiki-gigaword-50")
```

```python
#Choose words
words = ["king", "queen", "diamond"]

for w in words:
    print(f"\nWord: {w}")
    print("First 10 vector values:", glove[w][:10])
    print("\nMost similar words:")
    print(glove.most_similar(w, topn=5))
    print("-" * 50)
```

```python
#5. Sentence-Level Embeddings
#Create sentences
sentences = [
    "AI is transforming businesses.",
    "Machine learning improves decision making.",
    "Jewellery stores use data analytics.",
    "Diamonds are valuable gemstones.",
    "Neural networks can learn patterns."
]
```

```python
#Compute sentence vectors (average of word vectors)
import numpy as np

def sentence_embedding(sentence):
    vectors = []
    for word in sentence.lower().split():
        if word in glove:
            vectors.append(glove[word])
    return np.mean(vectors, axis=0)

sentence_vectors = [sentence_embedding(s) for s in sentences]
```

```python
#Similarity Matrix
from sklearn.metrics.pairwise import cosine_similarity

similarity_matrix = cosine_similarity(sentence_vectors)

print("Similarity Matrix:\n")
print(similarity_matrix)
```
